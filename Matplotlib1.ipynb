```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

BASIC

```python
x=np.linspace(0,5,11)
x
```

```python
y=x**2
y
```

```python
plt.xlabel('xlab')
plt.ylabel('ylab')
plt.title("first plot")
plt.plot(x,y)
```

SUBPLOTS

```python
##multiple plots in one canvas

plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x**2,y**2)
plt.subplot(2,2,2)
plt.plot(x,x)
```

OBJECT ORIENTED WAY

```python
fig=plt.figure(figsize=(12,12),dpi=500)
axes1=fig.add_axes([1,0.9,1,1])
axes2=fig.add_axes([0.1,0.3,0.1,0.1])
axes1.plot(x,y)
axes2.plot(x,y)
#here list contains dist from left,dist form bottom,width,height
```

SAVING FIG

```python
fig1=plt.figure()
px=fig1.add_axes([1,0.6,1,0.2])
px.plot(x,y)
plt.savefig("basicplot.png")
```
