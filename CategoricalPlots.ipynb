Categorical Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=sns.load_dataset('tips')
df
```

COUNTPLOT

```python
sns.countplot(x=df['sex'],hue=df['smoker'])
```

BARPLOT

```python
sns.barplot(x=df['sex'],y=df['total_bill'])
#categorical on x axis and continuous on y axis
```

```python
sns.barplot(x=df['sex'],y=df['total_bill'],estimator=np.sum)
#here estimator is by default mean
```

BOXPLOT

```python
sns.boxplot(x=df['tip'],y=df['day'],palette='rainbow')
##also sns.boxplot(x='tip',t='day',data=df)
```

VIOLIN PLOT

```python
sns.violinplot(x=df['day'],y=df['total_bill'])
```

STRIPPLOT

```python
sns.stripplot(x=df['day'],y=df['total_bill'])
```

SWARMPLOT

```python
sns.swarmplot(x='day',y='total_bill',data=df)
```

```python
#combining swarmplot and violinplot
sns.swarmplot(x='tip',y='day',data=df)
sns.violinplot(x=df['tip'],y=df['day'])
```
